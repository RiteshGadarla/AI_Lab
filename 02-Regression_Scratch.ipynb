```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([650, 800, 1200, 1500, 1850])
y = np.array([75, 100, 130, 160, 180])
```

```python
#Normalizing the data
X = (X - X.mean()) / X.std()
print(X)
```

```python
weights = 0
bias = 0
learning_rate = 0.01
epochs = 100
```

```python
for i in range(epochs):
    #Making Predictions
    predictions = weights * X + bias
    #Calculating Loss
    error = y - predictions
    loss = (error ** 2).mean()
    #calculate gradients
    gradient_weight = -2 * (X * error).mean()
    gradient_bias = -2 * error.mean()
    #updating parameters
    weights = weights - gradient_weight * learning_rate
    bias = bias - gradient_bias * learning_rate
    print("Weights: ", weights)
    print("Bias: ", bias)
```

```python
plt.scatter(X, y, label="Actual")
plt.plot(X, weights * X + bias, color='red', label="Predicted")
plt.title("Linear Regression from Scratch")
plt.xlabel("House Size")
plt.ylabel("House Price")
plt.legend()
plt.grid(True)
plt.show()
```
